```python
import pandas as pd

df = pd.read_json('shumei-label.jsonl', lines =True)
```

```python
df
```

```python
ffff = (df['phash']=="ffffffffffffffff")
df = df[~ffff]
```

```python
dfd = (
    df[df.duplicated(subset=["phash"], keep=False)]
    .sort_values(["pilmode"], ascending=False, kind="stable")
    .sort_values(["width"], ascending=False, kind="stable")
    .sort_values(["phash"], ascending=False, kind="stable")
)
dfd
```

```python
all_modes = df.sort_values(['pilmode'], ascending=False, kind='stable').drop_duplicates(subset=['pilmode'], keep='first')
all_modes
```

```python
df = (
    df.sort_values(["pilmode"], ascending=True, kind="stable")
    .sort_values(["width"], ascending=False, kind="stable")
    .sort_values(["phash"], ascending=False, kind="stable")

)
df
```

```python
import json
from pathlib import Path

root = Path("data/data_tool")

sections = []
count = {}
files = {}

data = {"sections": sections, "count": count, "files": files}

for c in cls:
    section = df[df["riskLabel1"] == c]
    section = section["img"].tolist()

    sections.append(c)

    count[c] = len(section)

    files[c] = section

json.dump(data, open("sections-shumei.json", "w"), indent=2, ensure_ascii=False)
```
