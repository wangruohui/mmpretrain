```python
classes = ["正常"]
classes_count = [0]
classes_two = ["正常"]
classes_two_count = [0]
classes_three = ["正常"]
classes_three_count = [0]

image_list = []
```

```python
from pathlib import Path

other_suffix = [".txt", ".md"]
im_suffix = [".jpg", ".jpeg", ".png", ".webp", ".bmp"]


def images_under(path: Path):
    for p in path.iterdir():
        if p.is_dir():
            yield from images_under(p)
        else:
            if p.suffix.lower() in im_suffix:
                yield p
            elif p.suffix.lower() not in other_suffix:
                print("unknown file:", p)


def get_label(name, lst, lst_count):
    if name not in lst:
        lst.append(name)
        lst_count.append(0)

    label = lst.index(name)
    assert label >= 0
    return label
```

```python
import pandas as pd
from os.path import expanduser

df = pd.read_excel(expanduser("~/data_tool/多模态图片拦截标准.xlsx"))

for i, row in df.iterrows():
    one = row["一级标签"]
    two = row["二级标签"]
    three = row["三级标签"]

    if one in ["色情", "性感"]:
        continue

    if one == '未成年人':
        if '色情' in two:
            one = '色情'
        elif '暴力' in two:
            one = '暴恐'
        else:
            one = '正常'

    imdir = f"res/{two}/google_{three}"
    imdir = imdir.replace("一号领导", "习近平")

    if not Path(imdir).exists():
        print("not exists:", imdir)
        continue

    if row["拦截标准"] == "不拦截":
        label = "正常"
        label_two = "正常"
    else:
        label = one
        label_two = two

    label = get_label(label, classes, classes_count)
    label_two = get_label(label_two, classes_two, classes_two_count)
    label_three = get_label(three, classes_three, classes_three_count)

    for img in images_under(Path(imdir)):
        image_list.append((str(img), label, label_two, label_three))
        classes_count[label] += 1
        classes_two_count[label_two] += 1
        classes_three_count[label_three] += 1
```

```python
from itertools import chain

cls_ = get_label("色情", classes, classes_count)
cls_two = get_label("色情", classes_two, classes_two_count)
cls_three = get_label("色情", classes_three, classes_three_count)

for f in chain(
    images_under(Path("safety240722/seqing/20240322_8k_nsfw_dup_1507")),
    images_under(Path("safety240722/seqing/nsfw_23k")),
):
    image_list.append((f, cls_, cls_two, cls_three))
    classes_count[cls_] += 1
    classes_two_count[cls_two] += 1
    classes_three_count[cls_three] += 1

print(classes)
print(classes_count)
print(classes_two)
print(classes_two_count)
print(classes_three)
print(classes_three_count)
```

```python
cls_ = get_label("性感", classes, classes_count)
cls_two = get_label("性感", classes_two, classes_two_count)
cls_three = get_label("性感", classes_three, classes_three_count)

for f in images_under(Path("raw_data/sexy")):
    image_list.append((str(f),cls_,cls_two,cls_three))
    classes_count[cls_] += 1
    classes_two_count[cls_two] += 1
    classes_three_count[cls_three] += 1

cls_ = get_label("正常", classes, classes_count)
cls_two = get_label("正常", classes_two, classes_two_count)
cls_three = get_label("正常", classes_three, classes_three_count)

for f in chain(images_under(Path("raw_data/neutral")), images_under(Path("raw_data/drawings"))):
    image_list.append((str(f),cls_,cls_two,cls_three))
    classes_count[cls_] += 1
    classes_two_count[cls_two] += 1
    classes_three_count[cls_three] += 1

print(classes)
print(classes_count)
print(classes_two)
print(classes_two_count)
print(classes_three)
print(classes_three_count)
```

```python
from pathlib import Path

root = Path("safety240813")

known_exts = [".txt", '.md']

cls_ = get_label("涉政", classes, classes_count)
cls_two = get_label("涉政", classes_two, classes_two_count)
cls_three = get_label("涉政", classes_three, classes_three_count)

cls_pos = get_label("正常", classes, classes_count)
cls_pos_two = get_label("正常", classes_two, classes_two_count)
cls_pos_three = get_label("正常", classes_three, classes_three_count)

for path in images_under(root):
    if "positive" in str(path):
        image_list.append((str(path), cls_pos, cls_pos_two, cls_pos_three))
        classes_count[cls_pos] += 1
        classes_two_count[cls_pos_two] += 1
        classes_three_count[cls_pos_three] += 1
    elif "negative" in str(path):
        image_list.append((str(path), cls_, cls_two, cls_three))
        classes_count[cls_] += 1
        classes_two_count[cls_two] += 1
        classes_three_count[cls_three] += 1
```
